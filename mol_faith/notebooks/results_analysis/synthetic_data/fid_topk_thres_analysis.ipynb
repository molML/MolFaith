# Fid+ by threshold (synthetic targets)

```python
from mol_faith.notebooks.results_analysis.shared import (
    SYNTHETIC_RESULTS,
    SYNTHETIC_TARGET_COLORS,
    SYNTHETIC_TARGETS,
    load_raw_data_all_thres,
    plot_fidelity_by_threshold,
    plot_fidelity_by_threshold_per_target,
)

OUTPUT_DIR = SYNTHETIC_RESULTS / 'result_plots' / 'fid_topk_thres_analysis'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
all_thres_data = [load_raw_data_all_thres(SYNTHETIC_RESULTS / folder, target_name)
                  for folder, target_name in SYNTHETIC_TARGETS.values()]
label_mapping = {target_name: label for label, (_, target_name) in SYNTHETIC_TARGETS.items()}
```

```python
plot_fidelity_by_threshold(all_thres_data, SYNTHETIC_TARGET_COLORS, label_mapping, figsize=(7.2, 4),
                           save_path=OUTPUT_DIR / 'fid_by_threshold_analysis.pdf');
```

```python
plot_fidelity_by_threshold_per_target(all_thres_data, label_mapping, figsize=(7.2, 4),
                                      save_path=OUTPUT_DIR / 'fid_by_threshold_analysis_per_target.pdf');
```
