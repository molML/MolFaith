```python
import pandas as pd
import math

csv_path = "path/to/multi_model_comparison.csv"

df = pd.read_csv(csv_path)

def to_power10(x, sigfigs=3):
    if x == 0:
        return "0"
    exp = int(math.floor(math.log10(abs(x))))
    coeff = x / (10 ** exp)
    return f"{coeff:.{sigfigs-1}f} × 10^{exp}"

formatted = df.copy()
for col in formatted.columns:
    if col != "model" and "pearson_r" not in col and "r2" not in col and "f3" not in col:
        formatted[col] = formatted[col].map(lambda v: to_power10(v, sigfigs=2))
    elif "pearson_r" in col or "r2" in col or "f3" in col:
        formatted[col] = formatted[col].map(lambda x: f"{x:.2}")

model_order = ['Transformer', 'CNN', 'GAT', 'GCN', 'GIN']
formatted = formatted.set_index('model').reindex(model_order)
formatted = formatted[['rmse_mean', 'rmse_std', 'mae_mean', 'mae_std', 'f3_mean', 'f3_std']]
print(formatted)
```
